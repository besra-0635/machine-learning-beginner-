```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Verimizi hazırlıyoruz (Pandas)
veri = {
    'Metrekare': [50, 70, 100, 120, 150],
    'Fiyat': [1500000, 2100000, 3000000, 3600000, 4500000]
}
df = pd.DataFrame(veri)
X = df[['Metrekare']]  # burada x her zaman bir dataframe olmalıdır
y=df['Fiyat']          #vector halindedir. tahmin etmeye çalıştığımız sey
model = LinearRegression()
model.fit(X,y)         # model öğrenildi
yeni_ev = [[90]]       # bu 90 m'2 lik evin fiyatını model tahmin edecek
tahmin = model.predict(yeni_ev)
print("yeni ev için fiyat tahmini:",tahmin)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
veri = {
    'Yas': [1, 3, 5, 8, 10],
    'Fiyat': [900000, 750000, 600000, 400000, 250000]
}
df = pd.DataFrame(veri)
X = df[['Yas']]
y = df['Fiyat']
model = LinearRegression()
model.fit(X,y)
yeni_car=[[6]]
tahmin = model.predict(yeni_car)
print("yeni araba için fiyat tahmini:", tahmin)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
veri = {
   'yaş': [10,4,6,8,12,14],
   'km': [200000,110000,50000,400000,200000,1000000],
   'fiyat': [380000, 620000, 680000, 250000, 310000, 120000]
}
df = pd.DataFrame(veri)
X = df[['yaş','km']]
y = df['fiyat']
model = LinearRegression()
model.fit(X,y)
garage_car= [[8,20000]]
tahmin = model.predict(garage_car)
print("predict for garage car:",tahmin)
```
